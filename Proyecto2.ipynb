```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering, DBSCAN, SpectralClustering
from sklearn.manifold import MDS
from sklearn.cluster import KMeans

names = [
    "kms11/cfz", "kms11/cfz", "kms11/cfz",
    "kms34/cfz", "kms34/cfz", "kms34/cfz",
    "kms11", "kms11","kms11",
    "kms34", "kms34", "kms34",
    "kms11/cfzMedia", "kms34/cfzMedia", "kms11Media", "kms34Media" #datos generados con mas medias
]

distance_matrix=euclidean_dist_matrix
# --------- 1. Agglomerative Clustering ------------
agglo = AgglomerativeClustering(n_clusters=4, metric='precomputed', linkage='average')
labels_agglo = agglo.fit_predict(distance_matrix)

# --------- 2. DBSCAN (con matriz de distancia) ------------
dbscan = DBSCAN(metric='precomputed', eps=30, min_samples=3)
labels_dbscan = dbscan.fit_predict(distance_matrix)

# --------- 3. Spectral Clustering (requiere similitud) ------------
# Convertir la distancia a similitud: similitud = 1 / (1 + distancia)
similarity_matrix = 1 / (1 + distance_matrix)
spectral = SpectralClustering(n_clusters=4, affinity='precomputed', random_state=2)
labels_spectral = spectral.fit_predict(similarity_matrix)

#----------------kmeans--------------------

kmeans = KMeans(n_clusters=4, random_state=42)
labels_kmeans = kmeans.fit_predict(distance_matrix)


# --------- Visualización (opcional) con MDS ------------
from sklearn.manifold import MDS

mds = MDS(n_components=2, dissimilarity='precomputed', random_state=4)
X_2D = mds.fit_transform(distance_matrix)



def plot_clusters(X, labels, title, names):
    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=X[:, 0], y=X[:, 1], hue=labels, palette='tab10', s=100, legend=False)

    # Anotar cada punto con su nombre (columna)
    for i, name in enumerate(names):
        plt.text(X[i, 0] + 0.5, X[i, 1], name, fontsize=9, ha='left', va='center')

    plt.title(title)
    plt.xlabel("MDS 1")
    plt.ylabel("MDS 2")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(title, format='pdf', bbox_inches='tight')
    plt.show()





plot_clusters(X_2D, labels_agglo, "Agglomerative Clustering",names)
plot_clusters(X_2D, labels_dbscan, "DBSCAN Clustering",names)
plot_clusters(X_2D, labels_spectral, "Spectral Clustering",names)
plot_clusters(X_2D, labels_kmeans, "kmean Clustering",names)


```
